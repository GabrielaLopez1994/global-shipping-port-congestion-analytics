# Business Views

## Objetivo

Crear vistas analíticas orientadas al negocio que simplifiquen el consumo de información desde Power BI.

Las vistas se construirán sobre las tablas Gold y permitirán responder preguntas estratégicas relacionadas al tráfico portuario mundial.

# Validación de Gold

Se verificará la existencia de todas las tablas necesarias para construir las vistas analíticas.

```python
spark.sql("""
SHOW TABLES IN workspace.gold
""").show(truncate=False)
```

Dado que se confirmó la existencia de las 4 tablas requeridas para la creación de las vistas analíticas, se procederá a crearlas.

# Business View 01: Country Ranking

Objetivo:

Ranking histórico de países según el volumen total de tráfico portuario.

```python
# Creación de la Vista 01
spark.sql("""

CREATE OR REPLACE VIEW
workspace.gold.vw_country_ranking

AS

SELECT

    c.country_name,

    SUM(
        f.port_traffic_teu
    ) AS total_teu

FROM workspace.gold.fact_port_traffic f

INNER JOIN workspace.gold.dim_country c

ON f.country_id = c.country_id

GROUP BY c.country_name

""")
```

```python
#Validación de la vista 01

display(

    spark.sql("""

    SELECT *

    FROM workspace.gold.vw_country_ranking

    ORDER BY total_teu DESC

    LIMIT 10

    """)

)
```

# Business View 02

Historical Growth

Objetivo:

Analizar la evolución anual del tráfico portuario mundial.

```python
# Creación de la vista 02
spark.sql("""

CREATE OR REPLACE VIEW
workspace.gold.vw_global_growth

AS

SELECT

    year,

    SUM(
        port_traffic_teu
    ) AS global_teu

FROM workspace.gold.fact_port_traffic

GROUP BY year

""")
```

```python
# Validación de la vista 02
display(

    spark.sql("""

    SELECT *

    FROM workspace.gold.vw_global_growth

    ORDER BY year

    """)

)
```

# Business View 03

Top Countries By Year

Objetivo:

Identificar los países líderes en tráfico portuario para cada periodo.%md


```python
# Creación de la vista 03
spark.sql("""

CREATE OR REPLACE VIEW
workspace.gold.vw_top_countries_by_year

AS

SELECT

    c.country_name,

    f.year,

    f.port_traffic_teu

FROM workspace.gold.fact_port_traffic f

INNER JOIN workspace.gold.dim_country c

ON f.country_id = c.country_id

""")
```

# Business View 04

Country Trend Analysis

Objetivo:

Facilitar análisis temporales por país.

```python
# Creación de la vista 04
spark.sql("""

CREATE OR REPLACE VIEW
workspace.gold.vw_country_trends

AS

SELECT

    c.country_name,

    f.year,

    f.port_traffic_teu

FROM workspace.gold.fact_port_traffic f

INNER JOIN workspace.gold.dim_country c

ON f.country_id = c.country_id

""")
```
